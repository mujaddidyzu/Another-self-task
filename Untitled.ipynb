```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([[1000], [1500], [2000], [2500], [3000]])
y = np.array([50, 65, 80, 95, 110])

model = LinearRegression()
model.fit(X, y)

predicted = model.predict(X)

print(f"intercept (β₀): {model.intercept_}")
print(f"dhal (β₁): {model.coef_[0]}")

plt.scatter(X, y, color='green', label='Real Data')
plt.plot(X, predicted, color='orange', label='Regression line')
plt.xlabel('Area (Square)')
plt.ylabel('Price (Lack Taka)')
plt.title('Area vs Price')
plt.legend()
plt.tight_layout()
plt.show()
```
